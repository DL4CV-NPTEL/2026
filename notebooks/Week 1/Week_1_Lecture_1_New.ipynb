## 1. What is computer vision?

> A field that seeks to automate and endow a computing framework with the ability to
> interpret images the way humans do. A sub-topic of Artificial Intelligence.

Look at an ordinary photograph and you can answer, effortlessly, questions like *where is
the gluestick?*, *what is the full title of that book?*, or *what is wrong with this
image?* The question this field asks is whether a machine can answer them too.

Other definitions from the literature, each emphasising something different:

| Definition | Source |
|---|---|
| "the construction of explicit, meaningful descriptions of physical objects from images" | Ballard and Brown, 1982 |
| "computing properties of the 3D world from one or more digital images" | Trucco and Verri, 1998 |
| "to make useful decisions about real physical objects and scenes based on sensed images" | Sockman and Shapiro, 2001 |

Notice the progression: from *describing objects*, to *recovering the 3D world*, to
*making decisions*. The field's sense of what counts as "vision" has widened over time.

## 3. Why is it hard?

**Vision is an inverse problem.** Physics and computer graphics work *forward*: given
geometry, materials, lighting and a camera, produce an image. Vision has to run that
backwards. From the image alone, with no knowledge of how the picture was taken or what
the camera settings were, we must recover shape, lighting, colour, objects and their
interactions. That inference is made from incomplete, partial and noisy information, and
in general it has no unique answer.

**The data is enormous.** Images are extremely high-dimensional, which makes the
computation heavy before any cleverness is added.

**Perception is not the same as measurement.** The human visual system does not report
pixel values, and a long list of illusions makes the point:

- the **Müller-Lyer illusion**: two lines of equal length look different,
- a variation of the **Hermann grid**: spots appear at intersections that are not there,
- **Adelson's brightness constancy illusion**: two patches of identical intensity look very different,
- counting the red Xs in two arrangements: identical content, very different difficulty.

**And we do not have a complete model to copy.** No full model of the human visual system
exists; the models we do have describe subsystems rather than the whole. Which parts of a
scene are *perceived* and which are *cognized*? When does an object matter for a task and
when does the context matter more? Computer vision is, in this sense, **AI-complete**:
solving it in general looks equivalent to solving artificial intelligence.
