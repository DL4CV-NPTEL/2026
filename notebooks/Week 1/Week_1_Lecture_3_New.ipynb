## 2. Models of reflection

When light strikes a surface, several things happen at once.

**Some is absorbed.** The fraction reflected is the **albedo** $\rho$; the fraction
absorbed is $1 - \rho$. Low-albedo surfaces absorb more.

**Some is reflected diffusely**, independent of viewing direction. Brick, cloth and rough
wood behave this way. **Lambert's cosine law** says the amount of reflected light is
proportional to $\cos\theta$, where $\theta$ is the angle of illumination. This is why
intensity depends on the illumination angle: less light arrives per unit area at oblique
angles.

**Some is reflected specularly**, and that part *does* depend on where you view it from.
A mirror is the extreme case. Most real surfaces have **both** diffuse and specular
components.

Other effects the simple model leaves out: transparency, refraction, subsurface
scattering, fluorescence and phosphorescence.

All of this is summarised by the **BRDF**, the Bidirectional Reflectance Distribution
Function: a model of local reflection that says how bright a surface appears when viewed
from one direction while light falls on it from another.

## 3. Colour

Light is composed of a **spectrum of wavelengths**. The coloured light arriving at the
sensor depends on two things multiplied together:

1. the colour of the **light source**, and
2. the colour of the **surface**.

This is why the same object photographs differently under daylight and tungsten light,
and why "what colour is this object" is not a question the sensor can answer on its own.

> **A question from the slides.** If the visible spectrum is VIBGYOR, why is colour
> represented with just **R, G and B**? The answer is not in the physics of light; it is
> in the three cone types of the human eye. Week 1 Lecture 4 and Week 2 Lecture 7 return
> to this.

## 4. From photons to pixels

**The Bayer grid.** A camera sensor measures intensity, not colour, so colour filters are
laid over the pixels in a **Bayer arrangement**: **50% green, 25% red, 25% blue**. Green
gets half the sites because human vision is most sensitive to it.

Each pixel therefore records only one of the three channels. **Demosaicing** algorithms
reconstruct the full-colour image by estimating the two missing values at every pixel from
its surrounding neighbours. Every photograph you have ever seen from a consumer camera is
the output of this interpolation.

**CCD versus CMOS.** Two ways of getting charge out of the sensor:

| | How it works |
|---|---|
| **CCD** | Photogenerated charge is moved from pixel to pixel to an output node, converted to voltage there, and an analog-to-digital converter turns each pixel into a digital value |
| **CMOS** | Charge is converted to voltage **inside each element**, using several transistors per pixel to amplify and move it |

## Key takeaways

| Concept | Key takeaway |
|---|---|
| **Four factors** | Light source, surface, sensor and representation together decide the image |
| **Albedo** | $\rho$ is the reflected fraction; $1 - \rho$ is absorbed |
| **Diffuse reflection** | Independent of viewing direction; follows Lambert's cosine law, $\propto \cos\theta$ |
| **Specular reflection** | Depends on viewing direction; most surfaces mix both kinds |
| **BRDF** | How bright a surface looks from one direction when lit from another |
| **Colour** | Light source colour times surface colour; the sensor cannot separate them alone |
| **Bayer grid** | 50% green, 25% red, 25% blue; one channel measured per pixel |
| **Demosaicing** | The other two channels at each pixel are interpolated from neighbours |
| **CCD vs CMOS** | Charge converted at an output node, versus inside each pixel element |
